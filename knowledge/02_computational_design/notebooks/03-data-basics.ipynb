```python
from compas.geometry import Box, Point
from compas_notebook import Viewer
from compas_notebook.config import Config

height = 1.5                          # one number [m]
coordinates = [0.0, 0.0, 0.0]         # an ordered list [m]
metadata = {"length_unit": "m", "author": "CAS student"}  # named fields
point = Point(*coordinates)
box = Box(2.0, 1.0, height, name="Study box")
print(type(height).__name__, type(coordinates).__name__, type(metadata).__name__)
print(box.name, "volume:", box.volume, "m³")
```

## Read a field, then change it
Use `metadata["length_unit"]` for a dictionary entry and `box.zsize` for a COMPAS
property. A point has `x`, `y` and `z` properties. Changing a property changes that
object. Metadata is a separate dictionary here; it is not automatically attached
to every COMPAS object.


```python
print("Unit:", metadata["length_unit"], "point x:", point.x)
original_volume = box.volume
box.zsize = 2 * height
point.x = 1.0
assert abs(box.volume - 2 * original_volume) < 1e-9
assert point.x == 1.0
print("After changing height:", box.volume, "m³")
print("Point coordinates:", list(point))

config = Config()
config.view.width = 700
config.view.height = 360
config.view.show_grid = False

config.view.camera.position = [4, -6, 3]
config.view.camera.target = [0, 0, 0]
view = Viewer(config=config)
view.scene.add(box.to_mesh(), color="#5552bb", show_edges=True)
view.show()
```
