```python
import numpy as np
from compas.geometry import Line, Pointcloud
from compas_notebook import Viewer
from compas_notebook.config import Config
```

## x + y = 3, x − y = 1

```python
A = np.array([[1.0, 1.0], [1.0, -1.0]])
b = np.array([3.0, 1.0])
```

## QR solve

```python
Q, R = np.linalg.qr(A)
xy = np.linalg.solve(R, Q.T @ b)
xy
```

## Scene

```python
config = Config()
config.view.width, config.view.height = 700, 360
config.view.camera.position = [4, -5, 6]
config.view.camera.target = [1.5, 1, 0]
viewer = Viewer(config=config)
viewer.scene.add(Line([0, b[0], 0], [b[0], 0, 0]), color="#5552bb")
viewer.scene.add(Line([0, -b[1], 0], [3, 3 - b[1], 0]), color="#e89b43")
viewer.scene.add(Pointcloud([[xy[0], xy[1], 0]]), color="#348f9e", pointsize=0.15)
viewer.show()
```
