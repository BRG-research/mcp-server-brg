```python
import sys

if sys.platform == "emscripten":
    import micropip
    await micropip.install([
        "https://ifcopenshell.github.io/wasm-wheels/ifcopenshell-0.8.4-cp313-cp313-pyodide_2025_0_wasm32.whl",
        "compas_model==0.9.3", "lark", "pydantic", "shapely", "typer",
    ])
    await micropip.install("compas_ifc==2.1.0", deps=False)
else:
    %pip install -q compas_ifc==2.1.0

from compas.geometry import Box, Frame
from compas_ifc.bim import BuildingInformationModel
from compas_notebook import Viewer
from compas_notebook.config import Config
```

## Wall in a storey

```python
model = BuildingInformationModel.template(storey_count=1, unit="m")
frame = Frame([2.0, 0.1, 1.5])
wall = model.create_wall(geometry=Box(4.0, 0.2, 3.0), frame=frame, name="Wall A", parent=model.storeys[0])
wall
```

## Save and load

```python
model.save("first-wall.ifc")
loaded = BuildingInformationModel("first-wall.ifc")
loaded.get_elements_by_type("IfcWall")
```

## Scene

```python
config = Config()
config.view.width, config.view.height = 700, 360
config.view.camera.position = [7, -8, 5]
config.view.camera.target = [2, 0, 1.5]
viewer = Viewer(config=config)
viewer.scene.add(Box(4.0, 0.2, 3.0, frame=frame).to_mesh(), color="#5552bb", show_edges=True)
viewer.show()
```
