```python
import compas
from compas_model.elements import BeamElement
from compas_model.models import Model
from compas_notebook import Viewer
from compas_notebook.config import Config
```

## Element in a group

```python
model = Model()
floor = model.add_group("Ground floor")
beam = model.add_element(BeamElement(width=0.2, depth=0.3, length=3.0, name="B1"), parent=floor)
list(model.elements())
```

## Scene

```python
config = Config()
config.view.width, config.view.height = 700, 360
config.view.camera.position = [4, -6, 3]
config.view.camera.target = [0, 0, 1.5]
viewer = Viewer(config=config)
viewer.scene.add(beam.modelgeometry, color="#5552bb", show_edges=True)
viewer.show()
```

## Save and load

```python
compas.json_dump(model, "one-beam.compas.json", pretty=True)
restored = compas.json_load("one-beam.compas.json")
list(restored.elements())
```
