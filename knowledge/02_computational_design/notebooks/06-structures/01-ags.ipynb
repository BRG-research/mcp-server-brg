```python
from compas_ags.ags import graphstatics
from compas_ags.diagrams import FormDiagram, FormGraph, ForceDiagram
from compas_notebook import Viewer
from compas_notebook.config import Config
```

## Form diagram

```python
lines = [
    [[3, 3, 0], [3.67, 5.01, 0]],
    [[6, -2, 0], [6, 0, 0]],
    [[0, -2, 0], [0, 0, 0]],
    [[0, 0, 0], [-2, 0, 0]],
    [[0, 0, 0], [3, 3, 0]],
    [[3, 3, 0], [6, 0, 0]],
    [[6, 0, 0], [0, 0, 0]],
]
form = FormDiagram.from_graph(FormGraph.from_lines(lines))
force = ForceDiagram.from_formdiagram(form)
```

## Force diagram

```python
form.edge_force(0, -3.0)
graphstatics.form_update_q_from_qind(form)
graphstatics.force_update_from_form(force, form)
form.edges_attribute("f")
```

## Scene

```python
config = Config()
config.view.width, config.view.height = 700, 360
config.view.camera.position = [6, 1.5, 14]
config.view.camera.target = [6, 1.5, 0]
viewer = Viewer(config=config)
viewer.scene.add(form, show_faces=False, show_vertices=True)
viewer.scene.add(force.translated([6, 0, 0]), show_faces=False, show_vertices=True)
viewer.show()
```
