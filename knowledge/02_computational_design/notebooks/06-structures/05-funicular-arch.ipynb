```python
from compas.geometry import Line, Point, Polyline, Vector
from compas_notebook import Viewer
from compas_notebook.config import Config
```

## Loads

```python
span, H = 10.0, 6.0
xs = [2.0, 4.0, 5.5, 8.0]
loads = [1.0, 2.0, 1.5, 1.0]
```

## Load line and pole

```python
loadline = [Point(12, 0, 0)]
for load in loads:
    loadline.append(loadline[-1] + Vector(0, 0, -load))
pole = Point(12 - H, 0, 0)
```

## Funicular from the rays

```python
def funicular(pole):
    slopes = [(point.z - pole.z) / (point.x - pole.x) for point in loadline]
    points, x = [Point(0, 0, 0)], 0.0
    for slope, xn in zip(slopes, xs + [span]):
        points.append(points[-1] + Vector(xn - x, 0, slope * (xn - x)))
        x = xn
    return points
```

## Move the pole for a horizontal closing string

```python
pole = pole + Vector(0, 0, funicular(pole)[-1].z / span * H)
arch = Polyline(funicular(pole))
rays = [Line(pole, point) for point in loadline]
[round(point.z, 3) for point in arch.points]
```

## Scene

```python
config = Config()
config.view.width, config.view.height = 700, 360
config.view.camera.position = [7, -16, 1]
config.view.camera.target = [7, 0, 1]
viewer = Viewer(config=config)
viewer.scene.add(arch, color="#5552bb")
viewer.scene.add(Polyline(loadline), color="#dc4b8d")
for ray in rays:
    viewer.scene.add(ray, color="#999999")
viewer.show()
```
