```python
import numpy as np
from compas.datastructures import Mesh
from compas.matrices import connectivity_matrix
from compas_notebook import Viewer
from compas_notebook.config import Config
```

## One quad

```python
mesh = Mesh()
for key, (x, y, z) in {10: [0, 0, 0], 20: [2, 0, 0], 30: [2, 1, 0], 40: [0, 1, 0]}.items():
    mesh.add_vertex(key=key, x=x, y=y, z=z)
face = mesh.add_face([10, 20, 30, 40])
mesh.face_vertices(face), mesh.face_normal(face), mesh.area()
```

## Connectivity matrix

```python
key_to_row = mesh.vertex_index()
edges = list(mesh.edges())
C = connectivity_matrix([(key_to_row[u], key_to_row[v]) for u, v in edges], rtype="array")
xyz = np.array([mesh.vertex_coordinates(key) for key in mesh.vertices()])
C, C @ xyz
```

## Scene

```python
config = Config()
config.view.width, config.view.height = 700, 360
config.view.camera.position = [4, -5, 4]
config.view.camera.target = [1, 0.5, 0]
viewer = Viewer(config=config)
viewer.scene.add(mesh, color="#b6cae7", show_edges=True, show_vertices=True)
viewer.show()
```
