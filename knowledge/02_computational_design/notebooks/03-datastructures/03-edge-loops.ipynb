```python
from compas.datastructures import Mesh
from compas_notebook import Viewer
from compas_notebook.config import Config
```

## Quad mesh

```python
mesh = Mesh.from_meshgrid(dx=9, nx=9)
```

## Loop and strip through one edge

```python
loop = mesh.edge_loop((34, 35))
edges, faces = mesh.edge_strip((34, 35), return_faces=True)
len(loop), len(faces)
```

## Scene

```python
config = Config()
config.view.width, config.view.height = 700, 360
config.view.camera.position = [4.5, -6, 10]
config.view.camera.target = [4.5, 4.5, 0]
viewer = Viewer(config=config)
viewer.scene.add(mesh, facecolor={face: "#e89b43" if face in faces else "#cccccc" for face in mesh.faces()}, show_edges=True)
for edge in loop:
    viewer.scene.add(mesh.edge_line(edge), color="#dc4b8d")
viewer.show()
```
