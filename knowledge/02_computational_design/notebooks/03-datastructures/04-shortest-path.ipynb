```python
import random
from compas.datastructures import Graph
from compas.geometry import KDTree, Pointcloud, Polyline
from compas_notebook import Viewer
from compas_notebook.config import Config
```

## Connect each point to its neighbours

```python
random.seed(1)
cloud = Pointcloud.from_bounds(10, 5, 3, 120)
tree = KDTree(cloud)
graph = Graph()
for i, point in enumerate(cloud):
    graph.add_node(i, x=point.x, y=point.y, z=point.z)
for i, point in enumerate(cloud):
    for _, j, _ in tree.nearest_neighbors(point, 4):
        if j != i and not graph.has_edge((i, j), directed=False):
            graph.add_edge(i, j)
graph.number_of_edges()
```

## Shortest path

```python
path = graph.shortest_path(0, 1)
route = Polyline([graph.node_coordinates(node) for node in path])
len(path), route.length
```

## Scene

```python
config = Config()
config.view.width, config.view.height = 700, 360
config.view.camera.position = [5, -8, 6]
config.view.camera.target = [5, 2.5, 1.5]
viewer = Viewer(config=config)
viewer.scene.add(graph, color="#5552bb", nodesize=0.08)
viewer.scene.add(route, color="#dc4b8d")
viewer.show()
```
