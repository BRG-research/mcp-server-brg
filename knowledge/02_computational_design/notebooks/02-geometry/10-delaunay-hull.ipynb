```python
from compas.datastructures import Mesh
from compas.geometry import Pointcloud, Polygon, convex_hull_xy, delaunay_triangulation
from compas_notebook import Viewer
from compas_notebook.config import Config
```

## Site points

```python
points = list(Pointcloud.from_bounds(10, 6, 0, 40))
```

## Triangulation and hull

```python
mesh = Mesh.from_vertices_and_faces(points, delaunay_triangulation(points))
hull = Polygon(convex_hull_xy(points))
mesh.number_of_faces(), hull.area
```

## Scene

```python
config = Config()
config.view.width, config.view.height = 700, 360
config.view.camera.position = [5, -2, 12]
config.view.camera.target = [5, 3, 0]
viewer = Viewer(config=config)
viewer.scene.add(mesh, color="#b6cae7", show_edges=True)
viewer.scene.add(hull, color="#e89b43")
viewer.show()
```
