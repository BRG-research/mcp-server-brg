```python
from compas.geometry import Box, Frame, Line, Point, Sphere
from compas_notebook import Viewer
from compas_notebook.config import Config
```

## Box, sphere, point and line

```python
box = Box(1, 2, 3, frame=Frame([0, 0, 1.5]))
sphere = Sphere(1.5, point=[4, 0, 1.5])
point = Point(7, 0, 0)
line = Line([9, 0, 0], [9, 0, 3])
```

## Scene

```python
config = Config()
config.view.width, config.view.height = 700, 360
config.view.camera.position = [4.5, -12, 6]
config.view.camera.target = [4.5, 0, 1.5]
viewer = Viewer(config=config)
viewer.scene.add(box.to_mesh(), color="#348f9e", show_edges=True)
viewer.scene.add(sphere.to_mesh(u=32, v=16), color="#dc4b8d")
viewer.scene.add(point, color="#5552bb", pointsize=0.3)
viewer.scene.add(line, color="#e89b43")
viewer.show()
```
