```python
from compas.geometry import Box, Frame, Pointcloud
from compas_notebook import Viewer
from compas_notebook.config import Config
```

## Point grid

```python
n_columns = 5
n_rows = 4
points = []
for row in range(n_rows):
    for column in range(n_columns):
        points.append([column * 0.5, 0, row * 0.25])
```

```python
config = Config()
config.view.width, config.view.height = 700, 360
config.view.camera.position = [3, -5, 2]
config.view.camera.target = [1, 0, 0.5]
viewer = Viewer(config=config)
viewer.scene.add(Pointcloud(points), color="#5552bb", pointsize=0.08)
viewer.show()
```

## Staggered bricks

```python
length, depth, height, gap = 0.40, 0.20, 0.15, 0.01
pitch = length + gap
bricks = []
for row in range(n_rows):
    offset = (row % 2) * pitch / 2
    for column in range(n_columns):
        point = [column * pitch + offset, 0, height / 2 + row * (height + gap)]
        bricks.append(Box(length, depth, height, frame=Frame(point)))
```

```python
viewer = Viewer(config=config)
for brick in bricks:
    viewer.scene.add(brick.to_mesh(), color="#c78558", show_edges=True)
viewer.show()
```
