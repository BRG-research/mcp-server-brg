```python
from math import radians
from compas.geometry import Box, Plane, Projection, Reflection, Rotation, Scale, Shear, Translation
from compas_notebook import Viewer
from compas_notebook.config import Config
```

## One cube

```python
mesh = Box(1).to_mesh().translated([0, 0, 0.5])
```

## Rotate, scale, mirror, shear, project

```python
transformations = [
    Rotation.from_axis_and_angle([0, 0, 1], radians(30)),
    Scale.from_factors([1, 2, 0.5]),
    Reflection.from_plane(Plane([0.75, 0, 0], [1, 0, 0])),
    Shear.from_angle_direction_plane(radians(30), [1, 0, 0], Plane.worldXY()),
    Projection.from_plane(Plane.worldXY()),
]
meshes = [mesh.transformed(Translation.from_vector([3 * i, 0, 0]) * T) for i, T in enumerate(transformations, start=1)]
```

## Scene

```python
config = Config()
config.view.width, config.view.height = 700, 360
config.view.camera.position = [8, -12, 6]
config.view.camera.target = [8, 0, 0.5]
viewer = Viewer(config=config)
viewer.scene.add(mesh, color="#5552bb", show_edges=True)
for transformed in meshes:
    viewer.scene.add(transformed, color="#e89b43", show_edges=True)
viewer.show()
```
