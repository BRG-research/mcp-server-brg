```python
from compas.geometry import Polygon, offset_polygon
from compas_notebook import Viewer
from compas_notebook.config import Config
```

## Regular polygons

```python
polygons = [Polygon.from_sides_and_radius_xy(n, 1.0).translated([3 * (n - 3), 0, 0]) for n in range(3, 9)]
[round(polygon.area, 2) for polygon in polygons]
```

## Offset

```python
offsets = [Polygon(offset_polygon(polygon.points, 0.2)) for polygon in polygons]
```

## Scene

```python
config = Config()
config.view.width, config.view.height = 700, 360
config.view.camera.position = [7.5, -4, 14]
config.view.camera.target = [7.5, 0, 0]
viewer = Viewer(config=config)
for polygon, offset in zip(polygons, offsets):
    viewer.scene.add(polygon, color="#5552bb")
    viewer.scene.add(offset, color="#e89b43")
    viewer.scene.add(polygon.centroid, color="#348f9e", pointsize=0.1)
viewer.show()
```
