```python
from compas.geometry import Bezier, Box, Frame, Polyline, Transformation
from compas_notebook import Viewer
from compas_notebook.config import Config
```

## Bezier curve

```python
curve = Bezier([[0, 0, 0], [3, 6, 0], [6, -3, 3], [10, 0, 0]])
frames = [curve.frame_at(t / 10) for t in range(11)]
```

## Place boxes

```python
box = Box(0.8, 0.4, 0.2).to_mesh()
boxes = [box.transformed(Transformation.from_frame_to_frame(Frame.worldXY(), frame)) for frame in frames]
```

## Scene

```python
config = Config()
config.view.width, config.view.height = 700, 360
config.view.camera.position = [5, -12, 8]
config.view.camera.target = [5, 1, 1]
viewer = Viewer(config=config)
viewer.scene.add(curve.to_polyline(50), color="#5552bb")
viewer.scene.add(Polyline(curve.points), color="#999999")
for box in boxes:
    viewer.scene.add(box, color="#e89b43", show_edges=True)
viewer.show()
```
