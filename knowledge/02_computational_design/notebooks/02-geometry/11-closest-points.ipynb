```python
from compas.geometry import KDTree, Line, Point, Pointcloud
from compas_notebook import Viewer
from compas_notebook.config import Config
```

## Points and target

```python
cloud = Pointcloud.from_bounds(10, 10, 0, 200)
target = Point(5, 5, 0)
```

## Ten nearest neighbours

```python
tree = KDTree(cloud)
nearest = tree.nearest_neighbors(target, 10, distance_sort=True)
[round(distance, 2) for _, _, distance in nearest]
```

## Scene

```python
config = Config()
config.view.width, config.view.height = 700, 360
config.view.camera.position = [5, -4, 12]
config.view.camera.target = [5, 5, 0]
viewer = Viewer(config=config)
viewer.scene.add(cloud, color="#5552bb", pointsize=0.1)
viewer.scene.add(target, color="#dc4b8d", pointsize=0.3)
for xyz, _, _ in nearest:
    viewer.scene.add(Line(target, xyz), color="#dc4b8d")
viewer.show()
```
