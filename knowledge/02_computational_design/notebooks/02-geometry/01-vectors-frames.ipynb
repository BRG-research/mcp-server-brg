```python
from compas.geometry import Box, Frame, Point, Transformation, Vector
from compas_notebook import Viewer
from compas_notebook.config import Config
```

## Dot and cross product

```python
u = Vector(2, 0, 0)
v = Vector(0, 3, 0)
w = u.cross(v)
u.dot(v), w, w.length
```

## Frame to frame

```python
frame = Frame([2, 1, 0], [0, 1, 0], [-1, 0, 0])
T = Transformation.from_frame_to_frame(Frame.worldXY(), frame)
Point(1, 0, 0).transformed(T)
```

## Place a box

```python
mesh = Box(1.2, 0.6, 0.4).to_mesh()
placed = mesh.transformed(T)
```

```python
config = Config()
config.view.width, config.view.height = 700, 360
config.view.camera.position = [6, -7, 5]
config.view.camera.target = [1, 0.5, 0]
viewer = Viewer(config=config)
viewer.scene.add(mesh, color="#5552bb", show_edges=True)
viewer.scene.add(placed, color="#e89b43", show_edges=True)
viewer.scene.add(Frame.worldXY())
viewer.scene.add(frame)
viewer.show()
```
