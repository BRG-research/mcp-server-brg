```python
from math import radians
from compas.geometry import Box, Pointcloud, Rotation, bounding_box, oriented_bounding_box_numpy
from compas_notebook import Viewer
from compas_notebook.config import Config
```

## Rotated point cloud

```python
cloud = Pointcloud.from_bounds(8, 5, 3, 100)
cloud.transform(Rotation.from_axis_and_angle([0, 0, 1], radians(30)) * Rotation.from_axis_and_angle([1, 0, 0], radians(15)))
```

## Axis-aligned and oriented box

```python
aabb = Box.from_bounding_box(bounding_box(cloud))
obb = Box.from_bounding_box(oriented_bounding_box_numpy(cloud))
aabb.volume, obb.volume
```

## Scene

```python
config = Config()
config.view.width, config.view.height = 700, 360
config.view.camera.position = [4, -12, 8]
config.view.camera.target = [3, 3, 2]
viewer = Viewer(config=config)
viewer.scene.add(cloud, color="#5552bb", pointsize=0.1)
viewer.scene.add(aabb.to_mesh(), color="#999999", opacity=0.2)
viewer.scene.add(obb.to_mesh(), color="#e89b43", opacity=0.5)
viewer.show()
```
