```python
from math import cos, sin
from compas.colors import ColorMap
from compas.datastructures import Mesh
from compas_notebook import Viewer
from compas_notebook.config import Config
```

## Wavy roof

```python
mesh = Mesh.from_meshgrid(dx=10, nx=20)
for vertex in mesh.vertices():
    x, y, _ = mesh.vertex_coordinates(vertex)
    mesh.vertex_attribute(vertex, "z", sin(x / 2) * cos(y / 2))
```

## Face colour from height

```python
heights = {face: mesh.face_center(face)[2] for face in mesh.faces()}
zmin, zmax = min(heights.values()), max(heights.values())
cmap = ColorMap.from_mpl("viridis")
facecolor = {face: cmap(z, zmin, zmax) for face, z in heights.items()}
zmin, zmax
```

## Scene

```python
config = Config()
config.view.width, config.view.height = 700, 360
config.view.camera.position = [5, -8, 8]
config.view.camera.target = [5, 5, 0]
viewer = Viewer(config=config)
viewer.scene.add(mesh, facecolor=facecolor, show_edges=False)
viewer.show()
```
