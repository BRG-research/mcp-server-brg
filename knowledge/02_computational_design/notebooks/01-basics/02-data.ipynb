```python
from compas.geometry import Box, Point
from compas_notebook import Viewer
from compas_notebook.config import Config
```

## Values

```python
height = 1.5
coordinates = [0.0, 0.0, 0.0]
metadata = {"length_unit": "m", "author": "CAS student"}
type(height), type(coordinates), type(metadata)
```

## Objects

```python
point = Point(*coordinates)
box = Box(2.0, 1.0, height, name="Study box")
box.name, box.volume
```

## Change properties

```python
box.zsize = 2 * height
point.x = 1.0
metadata["length_unit"], box.volume, point
```

## Copy

```python
trial = box.copy()
trial.xsize = box.xsize / 2
trial.volume, box.volume
```

## Scene

```python
config = Config()
config.view.width, config.view.height = 700, 360
config.view.camera.position = [4, -6, 3]
viewer = Viewer(config=config)
viewer.scene.add(box.to_mesh(), color="#5552bb", show_edges=True)
viewer.show()
```
