```python
import compas
from compas.geometry import Box, Point
from compas_notebook import Viewer
from compas_notebook.config import Config
```

## Session

```python
session = {
    "length_unit": "m",
    "box": Box(2.0, 1.0, 1.5, name="Study box"),
    "reference_point": Point(1, 0, 0),
}
```

## Save and load

```python
compas.json_dump(session, "box-study.compas.json", pretty=True)
restored = compas.json_load("box-study.compas.json")
restored
```

## Scene

```python
config = Config()
config.view.width, config.view.height = 700, 360
config.view.camera.position = [4, -6, 3]
viewer = Viewer(config=config)
viewer.scene.add(restored["box"].to_mesh(), color="#5552bb", show_edges=True)
viewer.show()
```
